# Continuação das inversões individuais e combinadas

Executa primeiro os casos individuais ainda pendentes e depois combina os arranjos por linha. L2 usa o eletrodo 67 como remoto/infinito. As pastas de saída devem ser novas.

```python
from pathlib import Path
import subprocess
import sys

ROOT = Path(r"D:\\WILLIAM\\DOLINA-ELETRICA")
SCRIPTS = ROOT / "06_SCRIPTS_PROCESSAMENTO"
INDIVIDUAL_SCRIPT = SCRIPTS / "resipy_individual_quad_synthetic.py"
COMBINED_SCRIPT = SCRIPTS / "resipy_combined_quad_synthetic.py"
INDIVIDUAL_OUTPUT = ROOT / "05_RESULTADOS_RESIPY" / "INVERSOES_INDIVIDUAIS_QUAD_CONTINUACAO"
COMBINED_OUTPUT = ROOT / "05_RESULTADOS_RESIPY" / "INVERSOES_COMBINADAS_QUAD_CAVIDADE"
INDIVIDUAL_CASES = ["L1_WS", "L2_DD", "L2_WS", "L2_PD"]
CONTACT_MAX_OHM = 40_000.0
PUBLISH_EACH_IMAGE = False  # True cria commit e push para cada PNG

for path in [INDIVIDUAL_SCRIPT, COMBINED_SCRIPT]:
    assert path.exists(), path
for path in [INDIVIDUAL_OUTPUT, COMBINED_OUTPUT]:
    if path.exists():
        raise FileExistsError(f"Escolha uma pasta de saída nova: {path}")
```

## 1. Casos individuais pendentes

```python
individual_command = [
    sys.executable, str(INDIVIDUAL_SCRIPT), str(ROOT), str(INDIVIDUAL_OUTPUT),
    "--contact-max-ohm", str(CONTACT_MAX_OHM),
    "--cases", *INDIVIDUAL_CASES,
]
if PUBLISH_EACH_IMAGE:
    individual_command.append("--publish-images")
subprocess.run(individual_command, check=True)
```

## 2. Arranjos combinados por linha

```python
combined_command = [
    sys.executable, str(COMBINED_SCRIPT), str(ROOT), str(COMBINED_OUTPUT),
    "--contact-max-ohm", str(CONTACT_MAX_OHM),
]
if PUBLISH_EACH_IMAGE:
    combined_command.append("--publish-images")
subprocess.run(combined_command, check=True)
```
