# Inversões individuais quadrilaterais e ensaio sintético

Executa casos selecionados com malha quadrilateral, corte de contato de 40 kΩ e `Cut Corners`. Para cada caso, o ResIPy gera um forward de cavidade com os mesmos quadrupolos reais e reinverte o sintético a partir de 5.000 Ω·m homogêneos.

Este notebook não foi executado e não retoma a pasta parcial anterior. Use sempre uma pasta de saída nova.

```python
from pathlib import Path
import json
import runpy
import subprocess
import sys

ROOT = Path(r"D:\WILLIAM\DOLINA-ELETRICA")
SCRIPTS = ROOT / "06_SCRIPTS_PROCESSAMENTO"
SCRIPT = SCRIPTS / "resipy_individual_quad_synthetic.py"
OUTPUT = ROOT / "05_RESULTADOS_RESIPY" / "INVERSOES_INDIVIDUAIS_QUAD_CAVIDADE_NOTEBOOK"
CASES = ["L1_DD", "L1_WS", "L2_DD", "L2_WS", "L2_PD"]
CONTACT_MAX_OHM = 40_000.0

assert SCRIPT.exists(), SCRIPT
if OUTPUT.exists():
    raise FileExistsError(f"Escolha uma pasta de saída nova: {OUTPUT}")
```

```python
module = runpy.run_path(str(SCRIPT), run_name="resipy_workflow_parameters")
print("Malha:")
print(json.dumps(module["MESH_PARAMETERS"], indent=2, ensure_ascii=False))
print("\nInversão:")
print(json.dumps(module["INVERSION_PARAMETERS"], indent=2, ensure_ascii=False))
print("\nSintético:")
print(json.dumps(module["SYNTHETIC_PARAMETERS"], indent=2, ensure_ascii=False))
```

## Selecionar casos

Edite `CASES` antes de executar. Para um piloto, por exemplo: `CASES = ["L1_DD"]`.

```python
valid_cases = {"L1_DD", "L1_WS", "L2_DD", "L2_WS", "L2_PD"}
unknown = set(CASES) - valid_cases
assert not unknown, f"Casos inválidos: {sorted(unknown)}"
print("Casos selecionados:", ", ".join(CASES))
print("Saída:", OUTPUT)
```

## Executar

A célula abaixo inicia somente os casos selecionados. Interromper o kernel também interrompe o subprocesso chamado pelo notebook.

```python
command = [
    sys.executable, str(SCRIPT), str(ROOT), str(OUTPUT),
    "--contact-max-ohm", str(CONTACT_MAX_OHM),
    "--cases", *CASES,
]
subprocess.run(command, check=True)
```
